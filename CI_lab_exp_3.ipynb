```python
import numpy as np

# Define fuzzy sets
A = np.array([0.2, 0.5, 0.8])
B = np.array([0.6, 0.4, 0.7])

# Cartesian product relation
R = np.minimum(A[:, None], B[None, :])

# Define relation S
S = np.array([
    [0.5, 0.7, 0.2],
    [0.4, 0.8, 0.6],
    [0.9, 0.3, 0.5]
])

# Create result matrix
T_min = np.zeros((R.shape[0], S.shape[1]))

# Max-Min composition using nested loops
for i in range(R.shape[0]):
    for j in range(S.shape[1]):

        values = []

        for k in range(R.shape[1]):
            minimum = min(R[i, k], S[k, j])
            values.append(minimum)

        T_min[i, j] = max(values)

# Display R
print("Relation R =")
print(R)

# Display S
print("\nRelation S =")
print(S)

# Display max-min result
print("\nMAX-MIN COMPOSITION")
print("T = R o S")
print(T_min)

# Display in table
print("\nTABULAR FORM")
print("       y1     y2     y3")

for i in range(T_min.shape[0]):
    print(f"x{i+1}   {T_min[i,0]:.2f}   "
          f"{T_min[i,1]:.2f}   "
          f"{T_min[i,2]:.2f}")
```

```python
import numpy as np

# Define fuzzy sets
A = np.array([0.2, 0.5, 0.8])
B = np.array([0.6, 0.4, 0.7])

# Cartesian product relation
R = np.minimum(A[:, None], B[None, :])

# Define relation S
S = np.array([
    [0.5, 0.7, 0.2],
    [0.4, 0.8, 0.6],
    [0.9, 0.3, 0.5]
])

# Create result matrices
T_min = np.zeros((R.shape[0], S.shape[1]))
T_product = np.zeros((R.shape[0], S.shape[1]))

# Max-Min composition
for i in range(R.shape[0]):
    for j in range(S.shape[1]):

        values = []

        for k in range(R.shape[1]):
            minimum = min(R[i, k], S[k, j])
            values.append(minimum)

        T_min[i, j] = max(values)

# Max-Product composition
for i in range(R.shape[0]):
    for j in range(S.shape[1]):

        values = []

        for k in range(R.shape[1]):
            product = R[i, k] * S[k, j]
            values.append(product)

        T_product[i, j] = max(values)

# Display matrices
print("MAX-MIN COMPOSITION")
print(T_min)

print("\nMAX-PRODUCT COMPOSITION")
print(T_product)

# Compare the matrices
print("\nCOMPARISON")
print("-" * 60)

for i in range(T_min.shape[0]):
    for j in range(T_min.shape[1]):

        if T_min[i, j] != T_product[i, j]:

            print(
                f"Entry ({i+1},{j+1}): "
                f"Max-Min = {T_min[i,j]:.2f}, "
                f"Max-Product = {T_product[i,j]:.2f}"
            )

print("\nAll differing entries have been displayed.")
```
