```python
import numpy as np
import matplotlib.pyplot as plt
import skfuzzy as fuzz

# -------------------------------------------------------------
# TASK 1: Universe of Discourse
# -------------------------------------------------------------
x = np.linspace(0, 10, 500)

# -------------------------------------------------------------
# TASK 2 & 3: Standard Membership Functions Plot
# -------------------------------------------------------------
tri = fuzz.trimf(x, [2, 5, 8])
trap = fuzz.trapmf(x, [1, 3, 6, 8])
gauss = fuzz.gaussmf(x, 5, 1.2)
gbell = fuzz.gbellmf(x, 2, 3, 5)
sig = fuzz.sigmf(x, 5, 2)

plt.figure(figsize=(10, 5))
plt.plot(x, tri, label='Triangular [2, 5, 8]')
plt.plot(x, trap, label='Trapezoidal [1, 3, 6, 8]')
plt.plot(x, gauss, label='Gaussian (mean=5, sigma=1.2)')
plt.plot(x, gbell, label='Gen. Bell (a=2, b=3, c=5)')
plt.plot(x, sig, label='Sigmoid (c=5, a=2)')
plt.title('Comparison of Standard Membership Functions')
plt.xlabel('Universe of Discourse (X)')
plt.ylabel('Membership Degree')
plt.ylim(-0.05, 1.05)
plt.grid(True)
plt.legend()
plt.show()

# -------------------------------------------------------------
# TASK 4: Triangular and Trapezoidal Variations
# -------------------------------------------------------------
tri_sym = fuzz.trimf(x, [2, 5, 8])
tri_asym = fuzz.trimf(x, [2, 3, 8])      # Shifted peak (asymmetric)
trap_narrow = fuzz.trapmf(x, [3, 4, 5, 6])
trap_wide = fuzz.trapmf(x, [1, 3, 7, 9])  # Wider core and support

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(x, tri_sym, label='Symmetric [2, 5, 8]')
axes[0].plot(x, tri_asym, '--', label='Asymmetric [2, 3, 8]')
axes[0].set_title('Triangular: Symmetry & Peak Variation')
axes[0].grid(True)
axes[0].legend()

axes[1].plot(x, trap_narrow, label='Narrow Core [3, 4, 5, 6]')
axes[1].plot(x, trap_wide, '--', label='Wide Core [1, 3, 7, 9]')
axes[1].set_title('Trapezoidal: Width & Core Variation')
axes[1].grid(True)
axes[1].legend()
plt.show()

# -------------------------------------------------------------
# TASK 5: Gaussian Variations (Fixed Mean, Varying Spread)
# -------------------------------------------------------------
gauss_s1 = fuzz.gaussmf(x, 5, 0.5)
gauss_s2 = fuzz.gaussmf(x, 5, 1.0)
gauss_s3 = fuzz.gaussmf(x, 5, 1.8)

plt.figure(figsize=(9, 4))
plt.plot(x, gauss_s1, label=r'$\sigma = 0.5$')
plt.plot(x, gauss_s2, label=r'$\sigma = 1.0$')
plt.plot(x, gauss_s3, label=r'$\sigma = 1.8$')
plt.title(r'Gaussian MF: Varying Spread ($\sigma$) at $\mu = 5$')
plt.xlabel('X')
plt.ylabel('Membership Degree')
plt.grid(True)
plt.legend()
plt.show()

# -------------------------------------------------------------
# TASK 6: Generalized Bell and Sigmoidal Parameter Variations
# -------------------------------------------------------------
gbell_slope1 = fuzz.gbellmf(x, 2, 1, 5)
gbell_slope2 = fuzz.gbellmf(x, 2, 4, 5)
sig_slope1 = fuzz.sigmf(x, 5, 0.8)
sig_slope2 = fuzz.sigmf(x, 5, 3.5)

fig, axes = plt.subplots(1, 2, figsize=(14, 4))
axes[0].plot(x, gbell_slope1, label='b = 1 (Smooth transition)')
axes[0].plot(x, gbell_slope2, '--', label='b = 4 (Steep slope)')
axes[0].set_title('Generalized Bell: Varying Parameter b')
axes[0].grid(True)
axes[0].legend()

axes[1].plot(x, sig_slope1, label='a = 0.8 (Gentle slope)')
axes[1].plot(x, sig_slope2, '--', label='a = 3.5 (Steep transition)')
axes[1].set_title('Sigmoidal: Varying Slope a')
axes[1].grid(True)
axes[1].legend()
plt.show()

# -------------------------------------------------------------
# TASK 7: Linguistic Sets (Low, Medium, High)
# -------------------------------------------------------------
low = fuzz.trapmf(x, [0, 0, 2, 4])
medium = fuzz.trimf(x, [2.5, 5, 7.5])
high = fuzz.trapmf(x, [6, 8, 10, 10])

plt.figure(figsize=(9, 4))
plt.plot(x, low, label='Low', color='blue')
plt.plot(x, medium, label='Medium', color='orange')
plt.plot(x, high, label='High', color='green')
plt.title('Fuzzy Sets: Low, Medium, High Partition')
plt.xlabel('X')
plt.ylabel('Membership Degree')
plt.grid(True)
plt.legend()
plt.show()

# -------------------------------------------------------------
# TASKS 8, 9 & 10: Property Computation & Subnormal Analysis
# -------------------------------------------------------------
def analyze_fuzzy_set(name, mf, x_space, eps=1e-3):
    height = np.max(mf)
    is_normal = np.isclose(height, 1.0, atol=1e-2)

    # Support: mf > 0
    support_idx = np.where(mf > eps)[0]
    support_range = (round(x_space[support_idx[0]], 2), round(x_space[support_idx[-1]], 2)) if len(support_idx) else None

    # Core: mf == height
    core_idx = np.where(np.isclose(mf, height, atol=1e-2))[0]
    core_range = (round(x_space[core_idx[0]], 2), round(x_space[core_idx[-1]], 2)) if len(core_idx) else None

    # Crossover points: zero crossings of (mf - 0.5)
    zero_crossings = np.where(np.diff(np.sign(mf - 0.5)))[0]
    crossover_pts = [round(float(x_space[idx]), 2) for idx in zero_crossings]

    # Boundary: 0 < mf < height
    boundary_idx = np.where((mf > eps) & (mf < height - 1e-2))[0]
    has_boundary = len(boundary_idx) > 0

    print(f"--- Properties for: {name} ---")
    print(f"Height:           {height:.3f}")
    print(f"Normal:           {is_normal}")
    print(f"Support Range:    {support_range}")
    print(f"Core Range:       {core_range}")
    print(f"Boundary Exists:  {has_boundary}")
    print(f"Crossover Points: {crossover_pts}\n")

# Normal and Subnormal Trapezoid
trap_normal = fuzz.trapmf(x, [1, 3, 6, 8])
trap_subnormal = 0.7 * trap_normal
gauss_test = fuzz.gaussmf(x, 5, 1.2)

analyze_fuzzy_set("Normal Trapezoidal", trap_normal, x)
analyze_fuzzy_set("Subnormal Trapezoidal", trap_subnormal, x)
analyze_fuzzy_set("Gaussian MF (Mean=5, Sigma=1.2)", gauss_test, x)

# Plot Normal vs Subnormal
plt.figure(figsize=(8, 4))
plt.plot(x, trap_normal, label='Normal Set (Height = 1.0)')
plt.plot(x, trap_subnormal, '--', label='Subnormal Set (Height = 0.7)')
plt.title('Comparison of Normal vs Subnormal Fuzzy Sets')
plt.xlabel('X')
plt.ylabel('Membership Degree')
plt.ylim(0, 1.1)
plt.grid(True)
plt.legend()
plt.show()
```
