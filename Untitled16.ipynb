```python
import numpy as np
import matplotlib.pyplot as plt
import os

# ---------------- DATASET A ----------------
np.random.seed(42)
cities = np.random.uniform(0, 100, (15, 2))
distance_A = np.rint(np.linalg.norm(
    cities[:, None] - cities[None, :], axis=2
)).astype(int)

# ---------------- DATASET B ----------------
# Try to load distance_matrix.txt. If not found, use a fallback dummy matrix.
try:
    distance_B = np.loadtxt('/content/distance_matrix.txt')
    distance_B = np.asarray(distance_B, dtype=float)
    has_dataset_B = True
except FileNotFoundError:
    print("Warning: '/content/distance_matrix.txt' not found.")
    print("Please upload the file to your Google Colab environment to run Dataset B.")
    distance_B = None
    has_dataset_B = False

# ---------------- PARAMETERS ----------------
ALPHA = 1.0
BETAS = [2, 3, 4, 5]
RHO = 0.10
Q = 100.0
Q0 = 0.9
TAU_MAX = 1.0
TAU_MIN = 0.01
ITERATIONS = 100

# ---------------- ACO ALGORITHM ----------------
def aco(distance, model, beta, seed=42):
    np.random.seed(seed)

    n = len(distance)
    ants = n

    pheromone = np.ones((n, n))
    heuristic = 1 / (distance + 1e-10)
    np.fill_diagonal(heuristic, 0)

    best_length = float('inf')
    best_tour = None
    convergence = []

    for iteration in range(ITERATIONS):

        tours = []
        lengths = []

        for ant in range(ants):
            start = ant % n
            tour = [start]
            unvisited = set(range(n))
            unvisited.remove(start)

            while unvisited:
                current = tour[-1]
                candidates = list(unvisited)

                weights = np.array([
                    (pheromone[current, j] ** ALPHA) *
                    (heuristic[current, j] ** beta)
                    for j in candidates
                ])

                if weights.sum() == 0:
                    probabilities = np.ones(len(candidates)) / len(candidates)
                else:
                    probabilities = weights / weights.sum()

                if model == 'ACS' and np.random.random() < Q0:
                    next_city = candidates[np.argmax(weights)]
                else:
                    next_city = np.random.choice(
                        candidates, p=probabilities
                    )

                # Online pheromone update
                if model == 'Density':
                    pheromone[current, next_city] += Q

                elif model == 'Quantity':
                    pheromone[current, next_city] += (
                        Q / distance[current, next_city]
                    )

                tour.append(next_city)
                unvisited.remove(next_city)

                # ACS local update
                if model == 'ACS':
                    pheromone[current, next_city] = (
                        (1 - RHO) * pheromone[current, next_city]
                        + RHO * 0.01
                    )
                    pheromone[next_city, current] = pheromone[current, next_city]

            tour.append(start)

            length = sum(
                distance[tour[i], tour[i+1]]
                for i in range(n)
            )

            tours.append(tour)
            lengths.append(length)

        iteration_best = np.argmin(lengths)

        if lengths[iteration_best] < best_length:
            best_length = lengths[iteration_best]
            best_tour = tours[iteration_best].copy()

        # Offline pheromone evaporation
        pheromone *= (1 - RHO)

        # Ant Cycle update
        if model == 'Cycle':
            for tour, length in zip(tours, lengths):
                for i in range(n):
                    a, b = tour[i], tour[i+1]
                    pheromone[a, b] += Q / length
                    pheromone[b, a] += Q / length

        # ACS global best update
        elif model == 'ACS':
            for i in range(n):
                a, b = best_tour[i], best_tour[i+1]
                pheromone[a, b] += RHO * Q / best_length
                pheromone[b, a] = pheromone[a, b]

        # MMAS global best update and bounds
        elif model == 'MMAS':
            for i in range(n):
                a, b = best_tour[i], best_tour[i+1]
                pheromone[a, b] += Q / best_length
                pheromone[b, a] = pheromone[a, b]

            pheromone = np.clip(
                pheromone, TAU_MIN, TAU_MAX
            )

        convergence.append(best_length)

    return best_length, best_tour, convergence


# ---------------- EXPERIMENT ----------------
models = ['Density', 'Quantity', 'Cycle', 'ACS', 'MMAS']

def evaluate_dataset(distance, dataset_name, optimal=None):

    results = {}

    plt.figure(figsize=(11, 6))

    for model in models:

        model_best = float('inf')
        model_curve = None
        model_tour = None

        for beta in BETAS:

            length, tour, curve = aco(
                distance, model, beta
            )

            if length < model_best:
                model_best = length
                model_curve = curve
                model_tour = tour

            print(
                dataset_name, model,
                'Beta:', beta,
                'Best length:', length
            )

        results[model] = (
            model_best, model_tour, model_curve
        )

        plt.plot(
            range(1, ITERATIONS + 1),
            model_curve,
            label=model
        )

    plt.xlabel('Iteration Number')
    plt.ylabel('Best Tour Length')
    plt.title('ACO Convergence - ' + dataset_name)
    plt.legend()
    plt.grid()
    plt.show()

    print('\nRESULTS:', dataset_name)

    for model, (length, tour, curve) in results.items():

        print('\nModel:', model)
        print('Best Tour:', tour)
        print('Best Tour Length:', length)

        if optimal is not None:
            deviation = abs(length - optimal) / optimal * 100
            print('Percentage Deviation:', deviation, '%')

    return results


# ---------------- RUN BOTH DATASETS ----------------
results_A = evaluate_dataset(
    distance_A, 'Dataset A'
)

if has_dataset_B:
    results_B = evaluate_dataset(
        distance_B, 'Dataset B', optimal=176
    )
else:
    print("\nSkipped Dataset B since '/content/distance_matrix.txt' is missing.")
```

```python

import numpy as np
import matplotlib.pyplot as plt

# ---------------- PARAMETERS ----------------
LAMBDA = 10
GENERATIONS = 200
SIGMA = 0.5
LOWER = -5.12
UPPER = 5.12
ADAPT_INTERVAL = 10
C = 0.82

np.random.seed(42)

# ---------------- RASTRIGIN FUNCTION ----------------
def rastrigin(x):
    return 20 + np.sum(
        x**2 - 10 * np.cos(2 * np.pi * x)
    )

# ---------------- INITIAL PARENT ----------------
parent = np.array([4.5, 4.5])
parent_fitness = rastrigin(parent)

success_count = 0
best_values = []
sigma_values = []

# ---------------- (1 + LAMBDA) ES ----------------
for generation in range(1, GENERATIONS + 1):

    offspring = []

    # Generate 10 offspring
    for i in range(LAMBDA):

        child = parent + np.random.normal(
            0, SIGMA, size=2
        )

        child = np.clip(child, LOWER, UPPER)

        offspring.append(child)

    # Evaluate offspring
    fitness = [rastrigin(child) for child in offspring]

    best_index = np.argmin(fitness)

    best_child = offspring[best_index]
    best_child_fitness = fitness[best_index]

    # Selection
    if best_child_fitness < parent_fitness:
        parent = best_child.copy()
        parent_fitness = best_child_fitness
        success_count += 1

    # Store convergence values
    best_values.append(parent_fitness)
    sigma_values.append(SIGMA)

    # 1/5th success rule every 10 generations
    if generation % ADAPT_INTERVAL == 0:

        success_rate = success_count / ADAPT_INTERVAL

        if success_rate > 0.2:
            SIGMA = SIGMA / C

        elif success_rate < 0.2:
            SIGMA = SIGMA * C

        success_count = 0

    # Print progress
    if generation % 20 == 0:
        print(
            'Generation:', generation,
            'Best Fitness:', parent_fitness,
            'Sigma:', SIGMA
        )

# ---------------- FINAL RESULTS ----------------
print('\nFINAL RESULTS')
print('Best Solution:', parent)
print('Minimum Function Value:', parent_fitness)
print('Final Mutation Step Size:', SIGMA)

# ---------------- CONVERGENCE GRAPH ----------------
plt.figure(figsize=(10, 5))
plt.plot(range(1, GENERATIONS + 1), best_values)
plt.xlabel('Generation')
plt.ylabel('Best Objective Function Value')
plt.title('Rastrigin Function Convergence - (1+10) ES')
plt.grid()
plt.show()

# ---------------- STEP SIZE GRAPH ----------------
plt.figure(figsize=(10, 5))
plt.plot(range(1, GENERATIONS + 1), sigma_values)
plt.xlabel('Generation')
plt.ylabel('Mutation Step Size (Sigma)')
plt.title('Mutation Step Size Adaptation')
plt.grid()
plt.show()
```
